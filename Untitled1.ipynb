```python
class State(TypedDict):
    email: str
    label: str
    result: str

# helper: ask Gemini and always get plain text back
def ask(p):
    c = llm.invoke(p).content
    return c if isinstance(c, str) else "".join(
        x.get("text", "") if isinstance(x, dict) else str(x) for x in c)

# Step 1: classify the email
def classify(state):
    label = ask("Classify this email with ONE word: SPAM, REPLY (needs a reply), "
                f"or FYI (just information).\n\n{state['email']}")
    return {"label": label.strip().upper()}

# Decision: which path next?
def route(state):
    if "SPAM" in state["label"]:
        return "spam"
    if "REPLY" in state["label"]:
        return "reply"
    return "fyi"

def handle_spam(state):
    return {"result": "Spam, ignored."}

def handle_fyi(state):
    return {"result": "FYI summary: " + ask(f"Summarize in one line:\n\n{state['email']}")}

# Reply path: 3 tones, you pick one
def handle_reply(state):
    email = state["email"]
    drafts = {
        "1": ask(f"Write a short, friendly, normal-toned reply to this email:\n\n{email}"),
        "2": ask(f"Write a short, polite reply that says NO and declines the request in this email:\n\n{email}"),
        "3": ask(f"Write a short, formal, professional reply to this email:\n\n{email}"),
    }
    print("1) NORMAL:\n", drafts["1"], "\n")
    print("2) NO (decline):\n", drafts["2"], "\n")
    print("3) PROFESSIONAL:\n", drafts["3"], "\n")

    choice = input("Pick 1, 2 or 3 (or press Enter to cancel): ")
    if choice in drafts:
        return {"result": f"Reply {choice} sent (pretend):\n{drafts[choice]}"}
    return {"result": "Cancelled"}

# Build the graph
g = StateGraph(State)
g.add_node("classify", classify)
g.add_node("spam", handle_spam)
g.add_node("fyi", handle_fyi)
g.add_node("reply", handle_reply)

g.add_edge(START, "classify")
g.add_conditional_edges("classify", route, {"spam": "spam", "fyi": "fyi", "reply": "reply"})
g.add_edge("spam", END)
g.add_edge("fyi", END)
g.add_edge("reply", END)
app = g.compile()

# Test email: change this text to try each path
email = "Hi Anjali, can we move tomorrow's meeting to 4 PM? Thanks, Rahul"

out = app.invoke({"email": email})
print("\nLabel:", out["label"])
print(out["result"])
```
